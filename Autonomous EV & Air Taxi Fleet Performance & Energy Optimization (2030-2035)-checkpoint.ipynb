# 5 questions on pandas

## Question 1: Multi-Aggregation & Efficiency Ranking

```python
# Grouping by Vehicle_Type and applying multiple aggregate functions
agg_df = aut.groupby("Vehicle_Type").agg(
    Avg_Energy_Consumed=("Energy_Consumed_kWh", "mean"),
    Std_Distance_Traveled=("Distance_Traveled_km", "std"),
    Max_Fare_USD=("Fare_Amount_USD", "max"),
    Avg_Energy_Efficiency=("Energy_Efficiency_km_per_kWh", "mean"),
)

# Calculating rank based on average energy efficiency
agg_df["Efficiency_Rank"] = (
    agg_df["Avg_Energy_Efficiency"]
    .rank(ascending=False, method="min")
    .astype(int)
)

print(agg_df)
```

## Question 2: Weather & Error Threshold Filtering via Group Transform

```python
# Filter for specific weather conditions
adverse_weather = aut[aut["Weather_Condition"].isin(["Stormy", "Rainy"])]

# Calculate the mean errors per vehicle type during adverse weather using transform
mean_errors = adverse_weather.groupby("Vehicle_Type")[
    "Autonomous_System_Errors"
].transform("mean")

# Identify vehicle types that meet the condition
valid_types = adverse_weather[mean_errors > 1.5]["Vehicle_Type"].unique()

# Filter original dataframe for these vehicle types
filtered_aut = aut[aut["Vehicle_Type"].isin(valid_types)]

print(f"Matching Vehicle Types: {valid_types}")
print(filtered_aut.head())
```

## Question 3: Quantile Binning & Complex Conditional Formatting

```python
# Binning Battery_Health_Pct into 4 quantile-based categories (qcut)
aut["Battery_Health_Bucket"] = pd.qcut(
    aut["Battery_Health_Pct"],
    q=4,
    labels=["Low", "Moderate", "High", "Optimal"],
)

# Vectorized condition for High_Risk classification
aut["High_Risk"] = (aut["Battery_Health_Bucket"] == "Low") & (
    aut["Autonomous_System_Errors"] >= 2
)

print(aut[["Vehicle_ID", "Battery_Health_Bucket", "Autonomous_System_Errors", "High_Risk"]].head(10))
```

## Question 4: Top-N Group-wise Selection & Pivot Summary

```python
# Top 2 vehicles per Weather_Condition by Distance_Traveled_km
top2_distance = (
    aut.sort_values(
        by=["Weather_Condition", "Distance_Traveled_km"], ascending=[True, False]
    )
    .groupby("Weather_Condition")
    .head(2)
)

print("--- Top 2 Distance Traveled per Weather Condition ---")
print(
    top2_distance[
        ["Weather_Condition", "Vehicle_ID", "Distance_Traveled_km"]
    ]
)

# Pivot Table for Average Fare Amount by Weather and Vehicle Type
pivot_fare = pd.pivot_table(
    aut,
    values="Fare_Amount_USD",
    index="Weather_Condition",
    columns="Vehicle_Type",
    aggfunc="mean",
).round(2)

print("\n--- Average Fare Amount Pivot Table ---")
print(pivot_fare)
```

## Question 5: Cumulative Outlier Detection via Z-score

```python
# Calculate Z-score of Energy Efficiency within each Vehicle_Type group
group_mean = aut.groupby("Vehicle_Type")[
    "Energy_Efficiency_km_per_kWh"
].transform("mean")
group_std = aut.groupby("Vehicle_Type")[
    "Energy_Efficiency_km_per_kWh"
].transform("std")

aut["Efficiency_ZScore"] = (
    aut["Energy_Efficiency_km_per_kWh"] - group_mean
) / group_std

# Filter outliers where |Z-score| > 2.5
outliers = aut[aut["Efficiency_ZScore"].abs() > 2.5]

print(f"Total Outliers Found: {len(outliers)}")
print(
    outliers[
        [
            "Vehicle_ID",
            "Vehicle_Type",
            "Energy_Efficiency_km_per_kWh",
            "Efficiency_ZScore",
        ]
    ]
)
```

# 7 questions on visualization

## Question 1: Custom Grouped Boxplot with Outlier Highlight

```python
plt.figure(figsize=(12, 7))

# Custom styling properties for fliers (outliers)
flierprops = dict(marker='D', markerfacecolor='red', markersize=6, linestyle='none', markeredgecolor='black')

sns.boxplot(
    data=aut,
    x="Weather_Condition",
    y="Battery_Health_Pct",
    hue="Vehicle_Type",
    palette="Set2",
    flierprops=flierprops
)

plt.title("Battery Health Distribution by Weather Condition & Vehicle Type", fontsize=14)
plt.xlabel("Weather Condition", fontsize=12)
plt.ylabel("Battery Health (%)", fontsize=12)
plt.legend(title="Vehicle Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Question 2: Heatmap of Bivariate Correlation Matrix

```python
import numpy as np

# Select numerical columns
num_cols = aut.select_dtypes(include=[np.number]).columns
corr_matrix = aut[num_cols].corr()

# Generate a mask for the upper triangle
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)

plt.title("Correlation Matrix of EV & Air Taxi Metrics", fontsize=14)
plt.show()
```

## Question 3: Multivariate Bubble Chart with Faceting

```python
g = sns.FacetGrid(aut, col="Vehicle_Type", height=5, aspect=1.1, sharey=True)

# Map scatter plot to grid
g.map_dataframe(
    sns.scatterplot,
    x="Distance_Traveled_km",
    y="Energy_Consumed_kWh",
    hue="Autonomous_System_Errors",
    size="Fare_Amount_USD",
    sizes=(20, 200),
    palette="viridis",
    alpha=0.7
)

g.add_legend(title="Legend")
g.set_axis_labels("Distance Traveled (km)", "Energy Consumed (kWh)")
g.fig.subplots_adjust(top=0.8)
g.fig.suptitle("Energy Consumption vs. Distance Traveled by Vehicle Type", fontsize=14)

plt.show()
```

## Question 4: Bivariate Kernel Density Estimation (KDE) Contour Map

```python
plt.figure(figsize=(9, 7))

# Overlay KDE contour map
sns.kdeplot(
    data=aut,
    x="Battery_Capacity_kWh",
    y="Fare_Amount_USD",
    fill=True,
    thresh=0.05,
    levels=10,
    cmap="mako",
    alpha=0.8
)

# Overlay subtle scatter points
sns.scatterplot(
    data=aut,
    x="Battery_Capacity_kWh",
    y="Fare_Amount_USD",
    color="white",
    s=10,
    alpha=0.3
)

plt.title("2D KDE Density: Battery Capacity vs. Fare Amount", fontsize=14)
plt.xlabel("Battery Capacity (kWh)", fontsize=12)
plt.ylabel("Fare Amount (USD)", fontsize=12)
plt.show()
```

## Question 7: Subplots Grid - Multi-Column Overview

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# কলামসমূহের তালিকা
cols = ['Battery_Capacity_kWh', 'Battery_Health_Pct', 'Distance_Traveled_km', 'Energy_Consumed_kWh']
colors = ['skyblue', 'salmon', 'lightgreen', 'purple']

# লুপের মাধ্যমেSubplots তৈরি
for i, col in enumerate(cols):
    ax = axes[i // 2, i % 2]
    sns.histplot(aut[col], kde=True, ax=ax, color=colors[i])
    ax.set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```
